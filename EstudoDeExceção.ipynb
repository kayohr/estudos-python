```python
texto = 'hoje e nossa segunda aula do bootcamp, bootcamp de python'

novo_texto = texto.replace(',', ' ')
palavras = novo_texto.split()

contagem_de_palavras = {
    'hoje': 1,
    'bootcamp': 2,
    'e': 1

}

for palavra in palavras:
    if palavra in contagem_de_palavras:
        contagem_de_palavras[palavra]
    else:
        contagem_de_palavras[palavra] = 1

print(contagem_de_palavras)
```

```python
nome_valido = False
salario_valido = False
bonus_valido = False
senha_correta = "12345"  # Define a senha correta

while nome_valido is not True:
    try:
        # nome = "Luciano"
        nome = input("Digite o nome: ")
        # Verifica se o nome está vazio
        if len(nome) == 0:
            raise ValueError("O nome não pode estar vazio.")
        # Verifica se há números no nome
        elif nome.isdigit():
            raise ValueError("O nome não deve conter números.")
        else:
            nome_valido = True
            print("Nome válido:", nome)
    except ValueError as e:
        print(e)
        #exit()
while salario_valido is not True:
    try:
        # salario = 2000
        salario = float(input("Digite o valor do seu salário: "))
        if salario < 0:
            print("Por favor, digite um valor positivo para o salário.")
        else:
            salario_valido = True
    except ValueError:
        print("Entrada inválida para o salário. Por favor, digite um número.")

while bonus_valido is not True:
    try:
        # bonus = 3.0
        bonus = float(input("Digite o valor do seu bônus: "))
        if bonus < 0:
            print("Por favor, digite um valor positivo para o bônus.")
        else:
            bonus_valido = True
    except ValueError:
        print("Entrada inválida para o bônus. Por favor, digite um número.")

bonus_recebido = 1000 + salario * bonus  # Exemplo simples de KPI

tentativas = 3  # Define o número de tentativas permitidas
senha_valida = False

while tentativas > 0 and not senha_valida:
    senha = input("Digite a senha para liberar as informações: ")
    if senha == senha_correta:
        senha_valida = True
        print(f"{nome}, seu salário é R${salario:.2f} e seu bônus final é R${bonus_recebido:.2f}.")
    else:
        tentativas -= 1
        print(f"Senha incorreta. Você tem mais {tentativas} tentativa(s).")

if not senha_valida:
    print("Número máximo de tentativas atingido. Reiniciando o processo.")
```
